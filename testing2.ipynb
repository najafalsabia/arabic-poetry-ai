# Task 2 — Testing on Multiple Poems

Runs `find_original_poem` + `rewrite_poem_validated` on several different poems/fragments in one go.

```python
import os
import pandas as pd
from rewrite_poem import find_original_poem, rewrite_poem_validated
```

```python
# Set your API key before running 
#os.environ["OPENAI_API_KEY"] = "sk-"

df = pd.read_csv("full_final_poems.csv")
print(f"Number of poems in dataset: {len(df)}")
```

## Test Cases

Each case: (a fragment the user recognizes, to identify the original poem) + (the new first verse they write)

```python
test_cases = [
    {
        "fragment": "قفا نبك من ذكرى حبيب",
        "modified_first_verse": "قفا نبك من ذكرى حبيب ومنزل *** ولكن على أطلال قلب مكبل",
    },
    {
        "fragment": "ألا رب يوم لك منهن صالح",
        "modified_first_verse": "ألا رب يوم لي به الروح فرحة *** وأخرى تئن النفس فيها وتصرخ",
    },
    {
        "fragment": "القلب أعلم",
        "modified_first_verse": "القلب أعلم يا عذول بأمره *** وأحق منك بصبره وقراره",
    },
    {
        "fragment": "هل غادر الشعراء",
        "modified_first_verse": "هل غادر العشاق من متكلمِ *** أم هل عرفت الحب قبل تكلمي",
    },
    {
        "fragment": "لولا فواكه أيلول",
        "modified_first_verse": "لولا فضائل هذا العصر ما اجتمعت *** أنواع فخرٍ ورقّ الطبع والماءُ",
    },
    {
        "fragment": "أراعي بلوغ الشيب",
        "modified_first_verse": "أراقب طيف الحب والحب دائبٌ *** وأقضي الليالي والليالي دنو بيا",
    },
]
```

## Run All Cases and Collect Results

This table feeds directly into the final evaluation report (results + retry count + rhyme consistency).

```python
results = []

for i, case in enumerate(test_cases, 1):
    print(f"=== Case {i}: {case['fragment']} ===")

    found = find_original_poem(case["fragment"], df=df)

    if found is None:
        print("❌ Original poem not found\n")
        results.append({
            "fragment": case["fragment"],
            "found": False,
            "title": None,
            "poet_name": None,
            "generated": None,
            "consistent": None,
            "attempts": None,
        })
        continue

    print(f"✅ Found: {found['title']} — {found['poet_name']} ({found['era']})")

    result = rewrite_poem_validated(
        found["full_text"],
        case["modified_first_verse"],
        found["poet_name"],
        found["era"],
    )

    print(result["text"])
    print(f"Attempts: {result['attempts']} | Rhyme consistent? {result['consistent']}")
    print()

    results.append({
        "fragment": case["fragment"],
        "found": True,
        "title": found["title"],
        "poet_name": found["poet_name"],
        "generated": result["text"],
        "consistent": result["consistent"],
        "attempts": result["attempts"],
    })
```

## Results Summary (feeds directly into the evaluation report)

```python
results_df = pd.DataFrame(results)
results_df
```

```python
# Overall success rate
found_count = results_df["found"].sum()
consistent_count = results_df["consistent"].sum()

print(f"Poems found: {found_count}/{len(results_df)}")
print(f"Rhyme consistent (first try or after retry): {consistent_count}/{found_count}")
```
